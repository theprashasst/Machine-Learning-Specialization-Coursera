```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X_train=np.array([[1.2,3.4,5.7],[3.2,5.4,7.7],[1.6,3.7,6.1]])
# X_train=X_train.T

X_train
```

```python
Y_train=np.array([[1,0,1]])
Y_train=Y_train.reshape(-1,1)
Y_train
```

```python
plot= plt.scatter(X_train[0].flatten(),Y_train.flatten(), c="r", marker="x")
plt.show()
```

```python
model= Sequential([
    Dense(units=4, activation='sigmoid',input_shape=(3,)),
    Dense(units=1, activation='sigmoid')
])
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.fit(X_train,Y_train)
```

```python
test_X=np.random.rand(1,3)
# test_X=np.array([[3.2,5.4,7.7]])
y_predict=model.predict(test_X)
print(test_X)
print(y_predict)
if y_predict[0] >0.5 : print(1)
else: print(0)
```
